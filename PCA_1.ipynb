**Principal component Analysis**

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
## Load the Dataaset
from sklearn.datasets import load_breast_cancer
````

````python
cancer_dataset=load_breast_cancer()
````

````python
print(cancer_dataset.DESCR)
````

````python
df=pd.DataFrame(cancer_dataset['data'], columns=cancer_dataset['feature_names'])
````


| Part | What It Does |
| :--- | :--- |
| `cancer_dataset['data']` | Extracts the feature values (numeric data) from the dataset |
| `cancer_dataset['feature_names']` | Extracts the column names (feature labels) |
| `pd.DataFrame(...)` | Creates a pandas DataFrame with the data and column names |
| `df` | Stores the DataFrame for further analysis |



````python
df.head()
````

````python
##STANDARDIZATION
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(df)
scaled_data=scaler.transform(df)
````

````python
scaled_data
# mean=0,std dev=1
````

````python
# Applying PCA
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
# pca.fit(scaled_data)
# pca.transform(scaled_data)
data_pca=pca.fit_transform(scaled_data)
````

````python
data_pca
````

````python
pca.explained_variance_
````

````python
##Visualization
plt.figure(figsize=(8,8))
plt.scatter(data_pca[:,0],data_pca[:,1],c=cancer_dataset['target'],cmap='plasma')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
````
